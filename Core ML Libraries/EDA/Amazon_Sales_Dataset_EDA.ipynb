```python
amazon_dataset['discounted_price'] = amazon_dataset['discounted_price'].str.replace('₹', '')
amazon_dataset['discounted_price'] = amazon_dataset['discounted_price'].str.replace(',', '')
amazon_dataset['discounted_price'] = amazon_dataset['discounted_price'].astype(float)

amazon_dataset['actual_price'] = amazon_dataset['actual_price'].str.replace('₹', '')
amazon_dataset['actual_price'] = amazon_dataset['actual_price'].str.replace(',', '')
amazon_dataset['actual_price'] =  amazon_dataset['actual_price'].astype(float)
```

```python
amazon_dataset['discount_percentage'] = amazon_dataset['discount_percentage'].str.replace('%', '').astype(float)
amazon_dataset['discount_percentage'] = amazon_dataset['discount_percentage'] / 100
```

```python
amazon_dataset['rating'].value_counts()
```

```python
amazon_dataset.query('rating == "|"')
```

```python
amazon_dataset['rating'] = amazon_dataset['rating'].str.replace('|', '3.9').astype(float)
```

```python
amazon_dataset['rating_count'] =amazon_dataset['rating_count'].str.replace(',', '').astype(float)
```

```python
amazon_dataset.info()
```

```python
amazon_dataset.describe()
```

```python
amazon_dataset.isnull().sum()
```

```python
amazon_dataset['rating_count'] = amazon_dataset['rating_count'].fillna(value=amazon_dataset['rating_count'].median())
```

```python
any_dubliactes = amazon_dataset.duplicated(subset=['product_id', 'product_name', 'category', 'discounted_price',
       'actual_price', 'discount_percentage', 'rating', 'rating_count',
       'about_product', 'user_id', 'user_name', 'review_id', 'review_title',
       'review_content', 'img_link', 'product_link']).any()
```

```python
sns.set_style('whitegrid')
plt.figure(figsize=(15, 10))
```

```python
sns.pairplot(amazon_dataset[['discounted_price', 'actual_price','rating']], diag_kind="kde")
plt.suptitle("Discounted Price vs Actual Price Relationship")
plt.show()
```

```python
feature_cols = ['discounted_price', 'actual_price', 'discount_percentage', 'rating', 'rating_count']
corr = amazon_dataset[feature_cols].corr(numeric_only=True)
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='rating', data=amazon_dataset)
plt.xticks(rotation=90)
plt.title('Product Ratings Distribution')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```
