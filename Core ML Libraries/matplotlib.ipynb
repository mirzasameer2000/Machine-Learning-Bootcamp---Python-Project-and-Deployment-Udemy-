```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 50)
```

```python
print(len(x))
```

```python
print(type(x))
```

```python
print(x)
```

```python
y = np.sin(x)
```

```python
print(y)
```

**1. Line Chart**

```python
plt.plot(x, y)
plt.title("Sin wave plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
plt.figure(figsize=(14, 4))
plt.plot(x, y, color="red", linestyle="--", marker="o", label="sin(x)")
plt.title("Sin wave plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid(True)
plt.legend()
plt.show()
```

**2. Bar Chart**

```python
students = ["Alice", "Bob", "Charlie", "David", "Eve"]
scores = [85, 92, 78, 90, 88]

plt.bar(students, scores, color="orange")
plt.title("Student Scores")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.show()
```

```python
plt.barh(students, scores, color="green")
plt.title("Student Scores (Horizontal)")
plt.xlabel("Scores")
plt.ylabel("Students")
plt.show()
```

**3. Scatter Plot**

```python
np.random.seed(42)

s = np.random.rand(100)
a = np.random.rand(100)

print(s[:5])
print(a[:5])
```

```python
plt.scatter(s, a, color="purple", marker="x", label="Data points", alpha=0.7)
plt.title("Scatter Plot")
plt.xlabel("Feature s")
plt.ylabel("Feature a")
plt.legend()
plt.show()
```

**4. Histogram**

```python
data = np.random.rand(1000)

plt.hist(data, bins=30, color="skyblue", edgecolor="black")
plt.title("Data Distribution (Histogram)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

**5. Multiple Plots (sub-plots)**

```python
z = np.linspace(0, 10, 100)

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.plot(z, np.sin(z), "r--")
plt.title("Subplot 1: Sin wave")

plt.subplot(1, 2, 2)
plt.plot(z, np.cos(z), "b--")
plt.title("Subplot 2: Cos wave")

plt.suptitle("Sin and Cos Waves")
plt.tight_layout()
plt.show()
```
