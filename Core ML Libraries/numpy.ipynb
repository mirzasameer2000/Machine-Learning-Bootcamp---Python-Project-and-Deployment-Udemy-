```python
arr = np.array([[10,20,30],
                [40,50,60],
                [70,80,90]])

print(arr[2,2]) # by indexing
print(arr[2]) # row
print(arr[:,1]) # column
```

```python
a = np.array([98,33,12])
b = np.array([34,63,84])

print(a)
print("________________________________")
print(b)
print("________________________________")

print("Addition :", a + b )
print("________________________________")
print("Subtraction :", a - b )
print("________________________________")
print("Multiplication :", a * b )
print("________________________________")
print("Division :", a / b )
print("________________________________")
```

```python
print("Dot Product", np.dot(a,b))
print("________________________________")
print("Cross Product", np.cross(a,b))
print("________________________________")
```

```python
arr_1 = np.arange(1,10)

reshaped = arr_1.reshape(3,3)

print(arr_1)
print("-----")
print(reshaped)
print("-----")
```

```python
flattened = reshaped.flatten()

print(flattened)
```
