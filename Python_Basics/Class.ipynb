**4. Class Attributes**

```python
class DOG:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def bark(self):
        print("Woof!")
```

```python
my_dog = DOG("Tyson", 3)
print(my_dog.name)
print(my_dog.age)
my_dog.bark()
```

**5. Demonstrate Inheritance**

```python
class MY_DOG:
    def __init__(self, name, age, breed):
        self.name = name
        self.age = age
        self.breed = breed
    def bark(self):
        print("Woof!")
```

```python
class puppy(MY_DOG):
    def play(self):
        print(self.name +" is playing.")
```

```python
puppy1 = puppy(name="Max", age=23, breed="Golden Retriever")
```

```python
print(puppy1.name)
print(puppy1.age)
print(puppy1.breed)
puppy1.bark()
puppy1.play()
```
