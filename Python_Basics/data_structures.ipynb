List - Ordered and Changable (Mutable)

```python
fruits = ["apple", "banana", "mango"]
print(fruits)
```

```python
print(fruits[0])
print(fruits[1])
```

```python
fruits[2] = "orange"  
print(fruits)
```

```python
a = [1, 2, 3, 4.6, "hello", True]
print(a)
```

```python
print(type(a))
```

```python
a.append("world")
print(a)
```

```python
a.remove(4.6)
print(a)
```

```python
print(len(a))
```

**loop through the list**

```python
for a in a:
    print(a)
```

Tuple - Ordered and unchangeable (Immutable)

```python
color = ("red", "green", "blue")
print(color)
print(type(color))
print(color[0])
```

```python
for i in color:
    print(i)
```

```python
x = [1, 2, 3, 4, 5]
y = tuple(x)

print(type(x))
print(type(y))

print(x)
print(y)
```

Set - Unordered and Unique

```python
students = {"John", "Emily", "Michael", "Sarah"}
print(students)
print(type(students))
```

```python
students.add("David")
print(students)
```

```python
students.remove("Emily")
print(students)
```

```python
for student in students:
    print(student)
```

```python
list_1 = [1, 2, 3, 3, 3, 3]
print(list_1)
```

```python
set_1 = set(list_1)
print(set_1)
```

```python
list_2 = [5, 4, 4, 4, 5, 6, 3, 63, 52,6,3,5,6,3,5,6,3,5,6,3,5,6]
print(list_2)
```

```python
set_2 = list(set(list_2))
for i in set_2:
    print(i)
```

DIsctionary -- Key-value data

```python
person = {"name": "John", "age": 30, "city": "New York", "is_student": False}
```

```python
person["name"]
```

```python
person['age'] = 26
print(person)
```

```python
print(type(person))
```

```python
for key,value in person.items():
    print(f"{key}: {value}")
```
