**2. Using try and except block**

```python
try: 
    num = int(input("Enter a number: "))
    print("You've entered: ", num)
except ValueError:
    print("That's not a valid number. Please try again.")
```

**3. Catching any error**

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
    print("Result of 10 divided by your number is: ", result)
except Exception as e:
    print("An error occurred :", str(e), "\nPlease try again.")
```

**4. Multiple except blocks**

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
    print("Result of 10 divided by your number is: ", result)

except ValueError:
    print("That's not a valid number. ")

except ZeroDivisionError:
    print("Cannot divide by zero. Please enter a non-zero number.")
```

**5. Finally block**

```python
try:
    num = int(input("Enter a number: "))
    print("You've entered: ", num)
except:
    print("An error occurred.")
finally:
    print("This block will always execute, regardless of whether an error occurred or not.")
```
