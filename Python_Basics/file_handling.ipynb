**1. Opening and Reading a file**

```python
file = open("E:\Machine Learning Bootcamp - Python, Project and Deployment - Udemy\cover.txt", "r")
```

```python
content = file.read()
print(content)

file.close()
```

**2. Using with to read (auto-closing)**

```python
with open("E:\Machine Learning Bootcamp - Python, Project and Deployment - Udemy\cover.txt", "r") as file:
    content = file.read()
    print(content)
```

**3. Writing to a File**

```python
with open("E:\Machine Learning Bootcamp - Python, Project and Deployment - Udemy\cover.txt", "a") as file:
    file.write("Hello, World!")
```

```python
with open("E:\Machine Learning Bootcamp - Python, Project and Deployment - Udemy\cover.txt", "r") as file:
    for line in file:
        print(line)
```
