1. For Loop

for 1 in sequence:

        run this code block

```python
print(range(5))
```

```python
print(*range(5))
```

```python
for i in range(0, 5):
    print('I am Data Scientist')
```

```python
for x in range(1, 6):
    print('Square of', x, 'is', x**2)
```

2. While loop

```python
count = 1

while count <= 5:
    print('Count is', count)
    count += 1
```

```python
while True:
    name = input('Enter your name: ')
    if name == 'exit':
        break
    print('Hello,', name)
```

3. Loop Controls - break;continue

```python
for i in range(0, 10):
    if i % 2 == 0:
        continue
    print(i)
```

```python
for g in range(0, 10):
    if g == 8:
        break
    else:
        print(g)
```

```python
for g in range(0, 10):
    if g == 8:
        continue
    else:
        print(g)
```
