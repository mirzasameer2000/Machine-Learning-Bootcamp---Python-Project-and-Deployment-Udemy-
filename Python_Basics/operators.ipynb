1. Arathemtic Operators

```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```

2. Comparison Operators

```python
x = 5
y = 10.2634

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

3. Logical Operators

```python
is_adult = True
has_id = False

print(is_adult and has_id)
print(is_adult or has_id)
print(not is_adult)
print(not has_id)
```

4. Assignment Operators

- = : Assign
- += Add and Assign
- -= Subtract and Assign
- *= Multiply and Assign
- /= Divide and Assign
- %= Modulus and Assign
- **= Exponent and Assign
- //= Floor Divide and Assign

```python
num = 10
print(num)

```

```python
num += 100
print(num)
```

```python
number = 7
number += 3
print(number)
number *= 2
print(number)
number -= 4
print(number)
number /= 2
print(number)
number %= 3
print(number)
number **= 2
print(number)
number //= 2
print(number)
```

5. Membership Operators
- in : True if value is found
- not in : True if value is not found

```python
f = [1, 2, 3, 4, 5]
print(3 in f)
print(6 in f)
print(3 not in f)
print(6 not in f)
```

6. Identity Operators 
- is : True if both refer to the same object
- is not : True if they do not refer to the same object

```python
q = [1, 2, 30, 4, 5]
w = [1, 2, 30, 4, 5]
c = q

print(q is w)
print(q is c)
print(c is q)

print(q is not w)
print(q is not c)
print(c is not q)
```
